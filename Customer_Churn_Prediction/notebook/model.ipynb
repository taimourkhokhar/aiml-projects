```python
cat_cols = [
    'gender',
    'Partner',
    'Dependents',
    'PhoneService',
    'MultipleLines',
    'InternetService',
    'OnlineSecurity',
    'OnlineBackup',
    'DeviceProtection',
    'TechSupport',
    'StreamingTV',
    'Contract',
    'PaperlessBilling',
    'PaymentMethod',
    'StreamingMovies'
]
```

```python
encoded_df = pd.get_dummies(df, columns=cat_cols, dtype=int)

encoded_df.head()
df=encoded_df
df.head()
```

```python
df.head()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
df = df.dropna(subset=['Churn'])
df['Churn'] = df['Churn'].map({'No': 0, 'Yes': 1})
x = df.drop('Churn', axis=1)
y = df['Churn']
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
scaler=StandardScaler()
```

```python
num_cols=['tenure','MonthlyCharges','TotalCharges']
x_train[num_cols]=scaler.fit_transform(x_train[num_cols])
x_test[num_cols] = scaler.transform(x_test[num_cols])
```

```python
from sklearn.linear_model import LogisticRegression
lg=LogisticRegression()
lg.fit(x_train,y_train)
```

```python
print(y_train.unique())
print(y_test.unique())
print(lg.predict(x_test[:5]))
print(type(y_test.iloc[0]))
```

```python
ypredictlg = lg.predict(x_test)

score = lg.score(x_test, y_test)
print(score)
print(ypredictlg[:10])
```

```python
from sklearn.model_selection import cross_val_score
scores=cross_val_score(lg,x,y,cv=5,scoring='accuracy')
scores
```

```python
from sklearn.metrics import ConfusionMatrixDisplay
cfd=ConfusionMatrixDisplay.from_estimator(lg, x_test, y_test)
cfd
```
